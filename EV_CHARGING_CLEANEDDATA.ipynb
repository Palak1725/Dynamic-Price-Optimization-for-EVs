```python
#Calculate session duration for both datasets
for df in [clean1, clean2]:
    df["session_duration_minutes"] = (
        df["disconnectTime"] - df["connectionTime"]
    ).dt.total_seconds() / 60


#Check negative and missing durations
for name, df in [
    ("Dataset 1", clean1),
    ("Dataset 2", clean2)
]:
    print(
        name,
        "Negative durations:",
        (df["session_duration_minutes"] < 0).sum()
    )

    print(
        name,
        "Missing durations:",
        df["session_duration_minutes"].isna().sum()
    )
```

```python
for df in [clean1, clean2]:

    df["idle_duration_minutes"] = (
        df["disconnectTime"] - df["doneChargingTime"]
    ).dt.total_seconds() / 60

    invalid_idle = (
        df["doneChargingTime"].notna()
        & (
            df["doneChargingTime"]
            > df["disconnectTime"]
        )
    )

    df.loc[
        invalid_idle,
        "idle_duration_minutes"
    ] = float("nan")
```

```python
for name, df in [
    ("Dataset 1", clean1),
    ("Dataset 2", clean2)
]:
    print("\n", name)

    print(
        "Missing idle durations:",
        df["idle_duration_minutes"].isna().sum()
    )

    print(
        "Negative idle durations:",
        (df["idle_duration_minutes"] < 0).sum()
    )
```

```python
for df in [clean1, clean2]:
    df["date"] = df["connectionTime"].dt.date
    df["hour"] = df["connectionTime"].dt.hour
    df["day_of_week"] = df["connectionTime"].dt.dayofweek
```

```python
def check_iqr_outliers(df, column):

    values = df[column].dropna()

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)

    iqr = q3 - q1

    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr

    outliers = (
        (df[column] < lower_limit)
        | (df[column] > upper_limit)
    )

    print("\nColumn:", column)
    print("Q1:", q1)
    print("Q3:", q3)
    print("IQR:", iqr)
    print("Lower limit:", lower_limit)
    print("Upper limit:", upper_limit)
    print("Outliers:", outliers.sum())

    return outliers
```

```python
for name, df in [
    ("Dataset 1", clean1),
    ("Dataset 2", clean2)
]:
    print("\n", name)

    df["duration_iqr_outlier"] = check_iqr_outliers(
        df, "session_duration_minutes"
    )

    df["energy_iqr_outlier"] = check_iqr_outliers(
        df, "kWhDelivered"
    )
```

```python
for df in [clean1, clean2]:

    for col in [
        "session_duration_minutes",
        "kWhDelivered"
    ]:

        mean_value = df[col].mean()
        std_value = df[col].std()

        if pd.notna(std_value) and std_value > 0:
            df[col + "_zscore"] = (
                df[col] - mean_value
            ) / std_value

        else:
            df[col + "_zscore"] = float("nan")
```

```python
for df in [clean1, clean2]:

    for col in [
        "session_duration_minutes",
        "kWhDelivered"
    ]:

        z_col = col + "_zscore"

        df[z_col + "_outlier"] = (
            df[z_col].abs() > 3
        )

        print(
            z_col,
            "potential outliers:",
            df[z_col + "_outlier"].sum()
        )
```
